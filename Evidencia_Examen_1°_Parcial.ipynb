Problema práctico de cálculo de RMSE: Se te dan los valores observados y predichos de una variable numérica para 4 observaciones: observados = [3, 7, 5, 10], predichos = [2, 8, 4, 9]. Calcula el RMSE

```python
import numpy as np

observados = np.array([3, 7, 5, 10])
predichos = np.array([2, 8, 4, 9])

# 1. Calcular los errores (residuales)
errores = observados - predichos

# 2. Elevar al cuadrado cada error
errores_cuadrados = errores**2

# 3. Calcular la media de los errores cuadrados (MSE)
mse = np.mean(errores_cuadrados)

# 4. Calcular la raíz cuadrada del MSE (RMSE)
rmse = np.sqrt(mse)

print(f"Valores observados: {observados}")
print(f"Valores predichos: {predichos}")
print(f"Errores: {errores}")
print(f"Errores al cuadrado: {errores_cuadrados}")
print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
```

**Problema de probabilidad condicional**

En una fábrica se producen piezas mediante dos máquinas, A y B.

La máquina A produce el 60% del total de las piezas.
La máquina B produce el 40% restante.
Sin embargo, no son igual de precisas:

La máquina A produce 2% de piezas defectuosas.
La máquina B produce 5% de piezas defectuosas.
Un inspector toma una pieza al azar del inventario general y encuentra que la pieza está defectuosa.

Pregunta:

¿Cuál es la probabilidad de que la pieza haya sido fabricada por la máquina B?

```python
# Datos proporcionados
P_A = 0.60  # Probabilidad de que una pieza sea de la máquina A
P_B = 0.40  # Probabilidad de que una pieza sea de la máquina B

P_D_dado_A = 0.02  # Probabilidad de que una pieza sea defectuosa de la máquina A
P_D_dado_B = 0.05  # Probabilidad de que una pieza sea defectuosa de la máquina B

# 1. Calcular la probabilidad total de que una pieza sea defectuosa (P(D))
P_D = (P_D_dado_A * P_A) + (P_D_dado_B * P_B)

# 2. Aplicar el Teorema de Bayes para encontrar P(B|D)
# P(B|D) = (P(D|B) * P(B)) / P(D)
P_B_dado_D = (P_D_dado_B * P_B) / P_D

print(f"Probabilidad total de que una pieza sea defectuosa (P(D)): {P_D:.4f}")
print(f"Probabilidad de que la pieza haya sido fabricada por la máquina B, dado que es defectuosa (P(B|D)): {P_B_dado_D:.4f}")
```

Acontinuación se te presenta una simulación de un caso en el que se pretende determinar si el consumo de café por los estudiantes ayuda a estudiar.

Imagina que seleccionas a 20 estudiantes al azar. A 10 les das café y a otros 10 les das agua, luegos les pides que resuelvan un examen de lógica.



```python
import pandas as pd
import numpy as np


# Establece una semilla para reproducibilidad
np.random.seed(42)


# 1. Crea una lista de matrículas de estudiantes
student_ids = [f'{i:06d}' for i in range(172120, 172120 + 20)]


# 2. Asigna 10 estudiantes al grupo 'Café' y 10 al grupo 'Agua'
groups = ['Café'] * 10 + ['Agua'] * 10
np.random.shuffle(groups)


# 3. Simula calificacionen de examenes para el grupo café
# Mean=70, Std=5, asegurate de que los valores estén entre (50-100)
coffee_scores = np.random.normal(loc=70, scale=5, size=10).round(0)
coffee_scores = np.clip(coffee_scores, 50, 100).astype(int)


# 4. Simula calificaciones de examen para el grupo de 'Agua'
# Mean=75, Std=5, asegurate de que los valores estén entre (50-100)
water_scores = np.random.normal(loc=75, scale=5, size=10).round(0)
water_scores = np.clip(water_scores, 50, 100).astype(int)


# Combina los valores basado en el grupo
exam_scores = []
coffee_idx = 0
water_idx = 0
for group in groups:
    if group == 'Café':
        exam_scores.append(coffee_scores[coffee_idx])
        coffee_idx += 1
    else:
        exam_scores.append(water_scores[water_idx])
        water_idx += 1


# 5. Combina en un DataFrame de Pandas
df_students = pd.DataFrame({
    'Matrícula': student_ids,
    'Grupo': groups,
    'Calificaciones': exam_scores
})


print("Datos simulados:")
print(df_students.head())

water_group = df_students[df_students['Grupo'] == 'Agua']
mean_water_score = water_group['Calificaciones'].mean()

print(f"La media de calificaciones para los estudiantes que consumieron agua es: {mean_water_score:.2f}")

coffee_group = df_students[df_students['Grupo'] == 'Café']
mean_coffee_score = coffee_group['Calificaciones'].mean()

print(f"La media de  los estudiantes que consumieron cafe es: {mean_coffee_score:.2f}")
```

```python
import pandas as pd

# Tabla de contingencia de curso vs. examen
data = {
    'Aprobó': [45, 20],
    'Reprobó': [5, 30]
}
df = pd.DataFrame(data, index=['Tomó el curso', 'No tomó el curso'])

# Valores necesarios para el cálculo
# Casos favorables
aprobo_y_tomo_curso = df.loc['Tomó el curso', 'Aprobó']

#  Total que tomaron el curso
total_tomo_curso = df.loc['Tomó el curso'].sum()

# Calcular la probabilidad condicional P(A|B)
probabilidad_condicional = aprobo_y_tomo_curso / total_tomo_curso

print("Tabla de Datos:")
print(df)
print("-" * 30)
print(f"Probabilidad de aprobar dado que tomó el curso P(A|B): {probabilidad_condicional:.2f}")
print(f"En porcentaje: {probabilidad_condicional * 100}%")
```

De la tabla de contingencia de la pregunta anterior calcula la probabilidad simple de aproba

```python
# Estudiantes que aprobaron
total_aprobaron = df['Aprobó'].sum()

# Total de estudiantes
total_estudiantes = df.sum().sum()

# Probabilidad simple de aprobar
probabilidad_simple_aprobar = total_aprobaron / total_estudiantes

print(f"Total de estudiantes que aprobaron: {total_aprobaron}")
print(f"Total de estudiantes: {total_estudiantes}")
print(f"La probabilidad simple de aprobar es: {probabilidad_simple_aprobar:.2f}")
```
