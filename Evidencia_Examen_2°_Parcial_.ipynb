```python
import pandas as pd
df_items = pd.read_csv('items.csv')
df_items = df_items.eval('importe = precio * cantidad')
print(df_items.head())
```

### Question 3

1. Lee ambos CSVs en dfc (clientes) y dfo (órdenes).
2. Realiza un left merge de dfc con dfo por cliente_id, manteniendo todas las filas de dfc.
3. Calcula:
(a) Número total de filas del resultado del left merge.
(b) Monto total por cliente (suma de monto) y reemplaza NaN por 0.

Pregunta numérica:

¿Cuál es el monto total por cliente cliente_id=1?

```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

dfc = pd.read_csv('clientes.csv')
dfo = pd.read_csv('ordenes.csv')

print('dfc head:')
print(dfc.head())
print('\ndfo head:')
print(dfo.head())
```

```python
merged_df = pd.merge(dfc, dfo, on='cliente_id', how='left')

print('Merged DataFrame head:')
print(merged_df.head())
print(f'Total number of rows in merged_df: {len(merged_df)}')
```

```python
total_monto_by_cliente = merged_df.groupby('cliente_id')['monto'].sum().fillna(0)
monto_cliente_1 = total_monto_by_cliente.loc[1]

print(f'Total monto for cliente_id=1: {monto_cliente_1}')
```

### Question 4

1. Carga el CSV en df, selecciona solo columnas: ts, temperatura, humedad.
2. Convierte ts a datetime manejando formatos mixtos (usa format='mixed' y errors='coerce').
3. Reemplaza valores faltantes en temperatura con la mediana y en humedad con la media (usa fillna).
4. Reporta:

(a) ¿Cuántos NaT quedaron en ts?

(b) Valor final de la mediana de temperatura tras imputación (redondea a 1 decimal).

(c) Valor final de la media de humedad tras imputación (redondea a 3 decimales).

```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

df = pd.read_csv('sensores.csv')
df = df[['ts', 'temperatura', 'humedad']]

print(df.head())
```

```python
df['ts'] = pd.to_datetime(df['ts'], format='mixed', errors='coerce')
nat_count = df['ts'].isna().sum()

print(f"Number of NaT values in 'ts' after conversion: {nat_count}")
```

```python
median_temperatura = df['temperatura'].median()
df['temperatura'] = df['temperatura'].fillna(median_temperatura)

mean_humedad = df['humedad'].mean()
df['humedad'] = df['humedad'].fillna(mean_humedad)

print("DataFrame after imputing missing values:")
print(df.head())
```

```python
final_median_temperatura = round(df['temperatura'].median(), 1)
final_mean_humedad = round(df['humedad'].mean(), 3)

print(f"Final median of temperatura after imputation: {final_median_temperatura}")
print(f"Final mean of humedad after imputation: {final_mean_humedad}")
```
