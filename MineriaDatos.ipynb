### Introduccion

#### Conceptos: La minería de datos (o data mining) es el proceso de analizar grandes volúmenes de datos para descubrir patrones, tendencias y correlaciones ocultas que no son evidentes a simple vista, transformando así datos brutos en conocimiento útil para la toma de decisiones estratégicas, predicción de resultados, optimización de procesos y aumento de ingresos. Utiliza técnicas estadísticas y de aprendizaje automático para encontrar anomalías y relaciones que permitan predecir comportamientos futuros, como la fuga de clientes o la compra conjunta de productos.


#### Ciencia de Datos: es un campo multidisciplinario que combina estadística, matemáticas, inteligencia artificial y computación para analizar grandes volúmenes de datos (Big Data) y extraer conocimiento significativo, patrones y tendencias, con el fin de apoyar la toma de decisiones, hacer predicciones y resolver problemas en diversas industrias. Implica todo el ciclo de vida de los datos, desde su recolección y limpieza hasta su análisis mediante técnicas como el Aprendizaje Automático (Machine Learning) y la visualización para generar información útil y estratégica.
