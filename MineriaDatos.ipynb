### Introduccion

#### Conceptos
La minería de datos (o data mining) es el proceso de analizar grandes volúmenes de datos para descubrir patrones, tendencias y correlaciones ocultas que no son evidentes a simple vista, transformando así datos brutos en conocimiento útil para la toma de decisiones estratégicas, predicción de resultados, optimización de procesos y aumento de ingresos. Utiliza técnicas estadísticas y de aprendizaje automático para encontrar anomalías y relaciones que permitan predecir comportamientos futuros, como la fuga de clientes o la compra conjunta de productos.


#### Ciencia de Datos
 Es un campo multidisciplinario que combina estadística, matemáticas, inteligencia artificial y computación para analizar grandes volúmenes de datos (Big Data) y extraer conocimiento significativo, patrones y tendencias, con el fin de apoyar la toma de decisiones, hacer predicciones y resolver problemas en diversas industrias. Implica todo el ciclo de vida de los datos, desde su recolección y limpieza hasta su análisis mediante técnicas como el Aprendizaje Automático (Machine Learning) y la visualización para generar información útil y estratégica.

#### Acceder a la documentación con ?

El lenguaje Python y su ecosistema de ciencia de datos se construyen pensando en el usuario, y una parte importante de ello es el acceso a la documentación.
Cada objeto de Python contiene una referencia a una cadena, conocida como *docstring*, que en la mayoría de los casos contendrá un resumen conciso del objeto y cómo usarlo.
Python tiene una función `help` integrada que puede acceder a esta información e imprimir los resultados.
Por ejemplo, para ver la documentación de la función `len` integrada, puede hacer lo siguiente:



```python
In [1]: help(len)
```

Debido a que encontrar ayuda sobre un objeto es tan común y útil, IPython y Jupyter introducen el carácter `?` como una abreviatura para acceder a esta documentación y otra información relevante:


```python
In [2]: len?
```

Esta notación funciona para casi cualquier cosa, incluidos los métodos de objeto:

```python
In [3]: L = [1, 2, 3]
In [4]: L.insert?
```

o incluso los propios objetos, con la documentación de su tipo:

```python
In [5]: L?
```

Es importante destacar que esto también funcionará con funciones u otros objetos que crees tú mismo.
Aquí definiremos una pequeña función con una cadena de documentación:

```python
In [6]: def square(a):
  ....:     """Return the square of a."""
  ....:     return a ** 2
  ....:
```

Ahora usaremos el signo de interrogación `?` para encontrar esta cadena de documentación:

1. Elemento de lista
2. Elemento de lista

```python
In [7]: square?
```

#### Accediendo al código fuente con ??

Dado que el lenguaje Python es tan legible, generalmente se puede obtener un mayor nivel de comprensión leyendo el código fuente del objeto que te interesa.
IPython y Jupyter ofrecen un acceso directo al código fuente con el doble signo de interrogación (`??`):

```python
In [8]: square??
```

Si experimentas con esto, notarás que a veces el sufijo `??` no muestra el código fuente: esto generalmente se debe a que el objeto en cuestión no está implementado en Python, sino en C o algún otro lenguaje de extensión compilado.
En este caso, el sufijo `??` genera el mismo resultado que el sufijo `?`.
Esto ocurre especialmente con muchos de los objetos y tipos integrados de Python, incluyendo la función `len` mencionada anteriormente:

```python
In [9]: len??
```

####Exploración de módulos con la tecla Tabulador

Otra interfaz útil es el uso de la tecla Tabulador para autocompletar y explorar el contenido de objetos, módulos y espacios de nombres.
En los siguientes ejemplos, usaré `<TAB>` para indicar cuándo se debe presionar la tecla Tabulador.

Completado de tabulaciones del contenido de objetos

Cada objeto de Python tiene varios atributos y métodos asociados.
Al igual que la función `help` mencionada anteriormente, Python incorpora una función `dir` que devuelve una lista de estos atributos, pero la interfaz de completado de tabulaciones es mucho más fácil de usar en la práctica.
Para ver una lista de todos los atributos disponibles de un objeto, puede escribir el nombre del objeto seguido de un punto ("`.`") y la tecla Tab.

```python
In [10]: dir(L)
```

Para limitar la lista, puede escribir el primer carácter o varios caracteres del nombre y la tecla Tab encontrará los atributos y métodos coincidentes:

```python
# ver atributos que empiezan con 'c'
In [10]: print([attr for attr in dir(L) if attr.startswith('c')])

# ver atributos que empiezan con 'co'
In [10]: print([attr for attr in dir(L) if attr.startswith('co')])
```

Si solo hay una opción, al presionar la tecla Tab se completará la línea automáticamente.
Por ejemplo, lo siguiente se reemplazará instantáneamente por `L.count`:

```python
In [10]: L.count
```

Aunque Python no establece una distinción estricta entre atributos públicos/externos y privados/internos, por convención se utiliza un guion bajo para indicar estos últimos.
Para mayor claridad, estos métodos privados y especiales se omiten de la lista por defecto, pero es posible listarlos escribiendo explícitamente el guion bajo:

```python
In [10]: print([attr for attr in dir(L) if attr.startswith('_')])
```

Completar tabulaciones al importar

El completado de tabulación también es útil al importar objetos desde paquetes.
Aquí lo usaremos para encontrar todas las posibles importaciones en el paquete `itertools` que comiencen por `co`:

```python
In [10]: import itertools
print([attr for attr in dir(itertools) if attr.startswith('co')])
```

De manera similar, puede usar la función de completar tabulaciones para ver qué importaciones están disponibles en su sistema (esto cambiará dependiendo de qué scripts y módulos de terceros sean visibles para su sesión de Python):

Más allá del autocompletado con tabulación: Coincidencia con comodines

El autocompletado con tabulación es útil si conoce los primeros caracteres del nombre del objeto o atributo que busca, pero no es de mucha ayuda si desea encontrar coincidencias en el medio o al final del nombre.
Para este caso, IPython y Jupyter ofrecen un método de coincidencia con comodines para nombres que usan el carácter `*`.

Por ejemplo, podemos usar esto para listar todos los objetos del espacio de nombres cuyo nombre termine en `Warning`:

```python
In [10]: *Warning?
```

Observe que el carácter `*` coincide con cualquier cadena, incluso la cadena vacía.

De forma similar, supongamos que buscamos un método de cadena que contiene la palabra `find` en su nombre.
Podemos buscarlo de esta manera:

```python
In [11]: str.*find*?
```

```python
# Cargar datos
df = pd.read_csv(StringIO(raw), sep='\t')
df.columns = [c.strip() for c in df.columns]
```

```python
# Variables
df['edad'] = pd.to_numeric(df['¿Cuántos años cumplidos tienes?'], errors='coerce')
map_sn = {'Si': True, 'No': False}
df['prog_python'] = df['¿Has realizado programas utilizando Python?'].map(map_sn)
df['usa_colab'] = df['¿Has utilizado Google Colab?'].map(map_sn)
df['cuenta_github'] = df['¿Tienes cuenta de GitHub?'].map(map_sn)
```

```python
# A NumPy
edad = df['edad'].to_numpy(dtype=float)
prog_py = df['prog_python'].to_numpy(dtype=bool)
usa_colab = df['usa_colab'].to_numpy(dtype=bool)
cuenta_gh = df['cuenta_github'].to_numpy(dtype=bool)
```

```python
# Descriptivos de edad (NumPy)
summary_age = {
    'n': int(edad.size),
    'min': float(np.min(edad)),
    'p25': float(np.percentile(edad, 25)),
    'mediana': float(np.median(edad)),
    'media': float(np.mean(edad)),
    'p75': float(np.percentile(edad, 75)),
    'max': float(np.max(edad)),
    'rango(ptp)': float(np.ptp(edad)),
    'var_muestral': float(np.var(edad, ddof=1)),
    'std_muestral': float(np.std(edad, ddof=1)),
    'cv': float(np.std(edad, ddof=1)/np.mean(edad)),
}
```

```python
# Frecuencias (NumPy)
freq_python = {'Si': int(np.count_nonzero(prog_py)), 'No': int(edad.size - np.count_nonzero(prog_py))}
freq_colab  = {'Si': int(np.count_nonzero(usa_colab)), 'No': int(edad.size - np.count_nonzero(usa_colab))}
freq_gh     = {'Si': int(np.count_nonzero(cuenta_gh)), 'No': int(edad.size - np.count_nonzero(cuenta_gh))}
```

```python
print(f"Frecuencia de utilizacion de Python: {freq_python}")
print(f"Frecuencia de utilizacion de Colab: {freq_colab}")
print(f"Frecuencia de utilizacion de Github: {freq_gh}")
```

```python
# Tablas cruzadas (NumPy)
def tabla2x2(a, b):
    t = np.zeros((2,2), dtype=int)
    t[1,1] = np.count_nonzero(a & b)
    t[1,0] = np.count_nonzero(a & ~b)
    t[0,1] = np.count_nonzero(~a & b)
    t[0,0] = np.count_nonzero(~a & ~b)
    return t

tab_py_colab = tabla2x2(prog_py, usa_colab)
tab_py_gh    = tabla2x2(prog_py, cuenta_gh)
tab_colab_gh = tabla2x2(usa_colab, cuenta_gh)
```
